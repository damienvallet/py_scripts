```python

# **********************************************************************
# Name : create_empty_mov.py
# Description : This script creates an dummy .mov video file with a specified duration and resolution.
# It generates a black video with a white rounds and lines appearing at random locations.                                                                    
# version : 1.0
# Author : Damien Vallet
# Licence : MIT License (MIT)
# Usage : python3 create_empty_mov.py
# ******************************************************************

# %pip install opencv-python-headless
import cv2
import numpy as np

def create_empty_mov(output_filename, duration_seconds, width=640, height=480, fps=24):
    """Creates an empty .mov video file.

    Args:
        output_filename: The name of the output .mov file (e.g., "output.mov").
        duration_seconds: The duration of the video in seconds.
        width: The width of the video frames in pixels.
        height: The height of the video frames in pixels.
        fps: Frames per second.
    """

    fourcc = cv2.VideoWriter_fourcc(*'M4S2')  # Codec 
    out = cv2.VideoWriter(output_filename, fourcc, fps, (width, height))

    # Create the video frames
    total_frames = int(duration_seconds * fps)
    black_frame = np.zeros((height, width, 3), dtype=np.uint8)
    white_pixel_pos = []

    for _ in range(total_frames):
        # Add a white pixel at a random location at a quarter of the frames
        x = np.random.randint(0, width)
        y = np.random.randint(0, height)
        white_pixel_pos.append((x, y))
        # draw line between the last 2 white pixels
        if len(white_pixel_pos) > 1:
            cv2.line(black_frame, white_pixel_pos[-1], white_pixel_pos[-2], (255, 255, 255), 1)
            cv2.ellipse(black_frame, white_pixel_pos[-1], (10, 10), 0, 0, 360, (255, 255, 255), -1)
        out.write(black_frame)
    out.release()

    print(f"Empty video '{output_filename}' created successfully with duration {duration_seconds} seconds.")


# Example usage
output_filename = "output.mov"
duration_seconds = 37
width = 1280
height = 720
fps = 24

create_empty_mov(output_filename, duration_seconds, width, height, fps)
```
